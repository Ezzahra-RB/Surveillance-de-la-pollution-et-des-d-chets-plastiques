```python
# ============================================================
# 1️⃣  Installation et importation des bibliothèques
# ============================================================

import torch
from ultralytics import YOLO
import os

print("✅ Torch version:", torch.__version__)
print("✅ CUDA disponible:", torch.cuda.is_available())
```

```python
# ============================================================
# 2️⃣  Définition du chemin de travail
# ============================================================
# IMPORTANT : adapte ce chemin à ton dossier local
base_dir = r"C:\Users\user\Documents\M-AI_S3\Computer_Vision\P_CV"
os.chdir(base_dir)
print("📁 Répertoire courant :", os.getcwd())

# Vérifie la présence du fichier data.yaml
assert os.path.exists("data.yaml"), "❌ Fichier data.yaml introuvable dans le dossier actuel !"
```

```python
# ============================================================
# 3️⃣  Vérification de la structure du dataset
# ============================================================
train_images = os.path.join(base_dir, "Taco_Dataset/train/images")
valid_images = os.path.join(base_dir, "Taco_Dataset/valid/images")
test_images = os.path.join(base_dir, "Taco_Dataset/test/images")

for path in [train_images, valid_images, test_images]:
    print(f"📂 {path} : {len(os.listdir(path))} fichiers")
```

```python
# ============================================================
# 4️⃣  Chargement du modèle YOLOv8 Medium pré-entraîné
# ============================================================
model = YOLO('yolov8m.pt')
print("✅ Modèle YOLOv8 Medium chargé avec succès.")
```

```python
# ============================================================
# 5️⃣  Entraînement (finetuning) avec augmentation
# ============================================================

results = model.train(
    data='data.yaml',       # fichier YAML de ton dataset
    epochs=100,              # nombre d'époques
    imgsz=640,              # taille des images
    batch=16,               # taille du batch
    degrees=20,             # rotation
    scale=0.5,              # changement d’échelle
    shear=5,
    perspective=0.0005,
    flipud=0.3,
    fliplr=0.5,
    # blur=0.3,               # flou
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    name='yolov8m_augmented', # nom du run
    project='runs',          # dossier de sortie
)
```

```python
# ============================================================
# 6️⃣  Visualisation des résultats d'entraînement
# ============================================================
# Le dossier de sortie contiendra :
# runs/train/yolov8m_augmented/
#   ├── results.png
#   ├── confusion_matrix.png
#   ├── F1_curve.png
#   ├── PR_curve.png
#   ├── labels.jpg
#   └── weights/
#        ├── best.pt
#        └── last.pt

from IPython.display import Image, display

# print("\n📊 Résultats de l'entraînement :")
# display(Image(filename='runs/train/yolov8m_augmented/results.png'))

print("\n📊 Résultats de l'entraînement :")
result_img = os.path.join("runs", "train", "yolov8m_augmented", "results.png")
if os.path.exists(result_img):
    display(Image(filename=result_img))
else:
    print("⚠️ L'image des résultats n'est pas encore disponible (attends la fin de l'entraînement).")
```

```python
# ============================================================
# 7️⃣  Évaluation du modèle sur le jeu de validation
# ============================================================
metrics = model.val()  # utilise le meilleur modèle automatiquement
print("\n📈 Résultats de la validation :")
print(metrics)
```

```python
# ============================================================
# 8️⃣  Prédiction sur le jeu de test
# ============================================================
# On teste sur quelques images du dossier test
test_results = model.predict(
    source="Taco_Dataset/test/images",  # chemin vers les images test
    conf=0.25,  # seuil de confiance
    save=True,  # sauvegarde les résultats dans runs/predict/
    name="taco_predictions"
)

print("✅ Prédictions sauvegardées dans : runs/predict/taco_predictions/")
```

```python
# ============================================================
# 9️⃣  Chargement du modèle finetuné pour utilisation future
# ============================================================
# (après entraînement, tu peux charger le meilleur modèle comme ceci)
model = YOLO('runs/train/yolov8m_augmented/weights/best.pt')
results = model.predict(source='path/to/your/image.jpg', conf=0.3)
```
