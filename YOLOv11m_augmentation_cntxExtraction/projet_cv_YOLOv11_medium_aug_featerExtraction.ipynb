```python
#===================================================
#6. MODULE D’EXTRACTION DE CONTEXTE : CNN encode un crop élargi autour des bboxes.
#===================================================
# context_encoder.py
import torch
import torch.nn as nn
import torchvision.transforms as T

class ContextEncoder(nn.Module):
    def __init__(self, out_dim=128):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Conv2d(3, 32, 3, 2, 1),
            nn.ReLU(),
            nn.Conv2d(32, 64, 3, 2, 1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d(1)
        )
        self.fc = nn.Linear(64, out_dim)

        self.transform = T.Compose([
            T.Resize((128, 128)),
            T.ToTensor()
        ])

    def forward(self, patch):
        x = self.encoder(patch)
        x = x.view(x.size(0), -1)
        return self.fc(x)


```

```python
#===================================================
#7. MODULE DE RECALIBRATION DES SCORES : 
#===================================================
# score_recalibrator.py
import torch
import torch.nn as nn

class ScoreRecalibrator(nn.Module):
    def __init__(self, yolo_dim=256, ctx_dim=128, n_classes=60):
        super().__init__()
        self.fc = nn.Sequential(
            nn.Linear(yolo_dim + ctx_dim, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, n_classes)
        )

    def forward(self, yolo_feat, ctx_feat):
        return self.fc(torch.cat([yolo_feat, ctx_feat], dim=1))

```

```python
#===================================================
#8. EXTRACTION DES FEATURES POUR ENTRAÎNER LE CLASSIFICATEUR
#===================================================
# extract_recalib_training_data.py

import os
import torch
from ultralytics import YOLO
from PIL import Image
from torchvision import transforms
import json

model = YOLO("runs/detect/train2/weights/best.pt")  # modèle finetuné

# transform = transforms.Compose([
#     transforms.Resize((224, 224)),
#     transforms.ToTensor()
# ])
transform = transforms.Compose([
    transforms.Resize((128,128)),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1),
    transforms.RandomRotation(degrees=10),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])


def enlarge_bbox(xyxy, scale=1.5, W=W, H=H):
    x1, y1, x2, y2 = xyxy
    w = x2 - x1
    h = y2 - y1
    cx = (x1 + x2) / 2
    cy = (y1 + y2) / 2

    new_w = w * scale
    new_h = h * scale

    nx1 = max(0, cx - new_w/2)
    ny1 = max(0, cy - new_h/2)
    nx2 = min(W, cx + new_w/2)
    ny2 = min(H, cy + new_h/2)

    return [nx1, ny1, nx2, ny2]

output = []

# img_dir = "images/train"
# label_dir = "labels/train"
# # train_images = os.path.join(base_dir, "Taco_Dataset/train/images")
img_dir = os.path.join(base_dir, "Taco_Dataset/train/images")
label_dir = os.path.join(base_dir, "Taco_Dataset/train/labels")

for img_file in os.listdir(img_dir):
    if not img_file.endswith(".jpg"):
        continue

    img_path = os.path.join(img_dir, img_file)
    img = Image.open(img_path).convert("RGB")
    W, H = img.size
    
    # prédictions YOLO
    results = model(img_path)[0]

    for box in results.boxes:
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        cls = int(box.cls[0])

        bbox = [x1, y1, x2, y2]
        bbox_ctx = enlarge_bbox(bbox, scale=1.3, W=W, H=H)

        crop = img.crop(bbox)
        crop_ctx = img.crop(bbox_ctx)

        crop.save(f"recalib/crops/{img_file}_{cls}.jpg")
        crop_ctx.save(f"recalib/context/{img_file}_{cls}.jpg")

        output.append([f"{img_file}_{cls}.jpg", cls])

with open("recalib/annotations.json", "w") as f:
    json.dump(output, f)

print("Extraction terminée.")
```

```python
#ENTRAÎNEMENT DU PETIT CLASSIFICATEUR

# train_recalibrator.py

import torch
from torch.utils.data import Dataset, DataLoader
#from context_encoder import ContextEncoder
#from score_recalibrator import ScoreRecalibrator
from PIL import Image
import json
import os

class RecalibDataset(Dataset):
    def __init__(self, json_path):
        self.items = json.load(open(json_path))
        self.transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor()
        ])
        self.ctx_encoder = ContextEncoder()  # shared weights

    def __len__(self):
        return len(self.items)

    def __getitem__(self, idx):
        filename, cls = self.items[idx]

        crop = Image.open(f"recalib/crops/{filename}")
        crop_ctx = Image.open(f"recalib/context/{filename}")

        crop = self.transform(crop)
        crop_ctx = self.ctx_encoder.transform(crop_ctx)

        ctx_feat = self.ctx_encoder(crop_ctx.unsqueeze(0)).squeeze(0)

        # YOLO feature = flattened crop (simplification)
        yolo_feat = crop.view(-1)[:256]

        return yolo_feat, ctx_feat, cls


dataset = RecalibDataset("recalib/annotations.json")
loader = DataLoader(dataset, batch_size=32, shuffle=True)

classifier = ScoreRecalibrator(256, 128, n_classes=60)
optimizer = torch.optim.AdamW(classifier.parameters(), lr=1e-4)
criterion = torch.nn.CrossEntropyLoss()

for epoch in range(15):
    for yolo_feat, ctx_feat, cls in loader:
        preds = classifier(yolo_feat, ctx_feat)
        loss = criterion(preds, cls)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch} - Loss {loss.item():.4f}")

torch.save(classifier.state_dict(), "recalib/classifier.pt")
print("🎯 Classificateur entraîné.")
```

```python
#INFERENCE AVEC SCORES RE-CALIBRÉS : Cette partie prend les détections YOLO 
#et les “corrige intelligemment” en utilisant le contexte pour rendre les scores plus fiables.

from ultralytics import YOLO
#from context_encoder import ContextEncoder
#from score_recalibrator import ScoreRecalibrator
from PIL import Image
import torch

model = YOLO("runs/detect/train2/weights/best.pt")

ctx_encoder = ContextEncoder()
classifier = ScoreRecalibrator()
classifier.load_state_dict(torch.load("recalib/classifier.pt"))

def infer(img_path):
    img = Image.open(img_path).convert("RGB")
    results = model(img_path)[0]

    for box in results.boxes:
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        cls = int(box.cls[0])
        score = float(box.conf[0])

        crop = img.crop((x1, y1, x2, y2))
        crop_ctx = img.crop(enlarge_bbox([x1,y1,x2,y2],1.3,img.width,img.height))

        yolo_feat = transforms.ToTensor()(crop).view(-1)[:256]
        ctx_feat = ctx_encoder(ctx_encoder.transform(crop_ctx).unsqueeze(0)).squeeze(0)

        new_scores = classifier(yolo_feat.unsqueeze(0), ctx_feat.unsqueeze(0))
        new_score = torch.softmax(new_scores, dim=1)[0][cls].item()

        print(f"Classe {cls} | Ancien score {score:.2f} → Nouveau score {new_score:.2f}")
```

```python
# ============================================================
# projet_cv_YOLOv11_medium_aug : Evaluation et inference
# ============================================================

import os
import glob
import torch
import json
import matplotlib.pyplot as plt
from ultralytics import YOLO
from PIL import Image
#from context_encoder import ContextEncoder
#from score_recalibrator import ScoreRecalibrator
from torchvision import transforms

# -------------------------------
# 1️⃣ Charger les modèles
# -------------------------------
yolo_model = YOLO('runs/detect/train2/weights/best.pt')
ctx_encoder = ContextEncoder()
classifier = ScoreRecalibrator()

# Charger classificateur recalibrateur si existant
classifier_path = "recalib/classifier.pt"
if os.path.exists(classifier_path):
    classifier.load_state_dict(torch.load(classifier_path))
    classifier.eval()
    print("✅ Recalibrateur chargé")
else:
    print("⚠️ Recalibrateur non trouvé. Les scores YOLO bruts seront utilisés.")

# -------------------------------
# 2️⃣ Visualisation de l'entraînement
# -------------------------------
run_dir = 'runs/yolov11m_augmented'
metrics_path = os.path.join(run_dir, 'metrics.json')

if os.path.exists(metrics_path):
    with open(metrics_path) as f:
        metrics = json.load(f)
    plt.figure(figsize=(12,5))
    
    # Loss
    plt.subplot(1,2,1)
    plt.plot(metrics['metrics']['box_loss'], label='Box Loss')
    plt.plot(metrics['metrics']['obj_loss'], label='Object Loss')
    plt.plot(metrics['metrics']['cls_loss'], label='Class Loss')
    plt.title('Loss during training')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    
    # mAP
    plt.subplot(1,2,2)
    plt.plot(metrics['metrics']['mAP_0.5'], label='mAP@0.5')
    plt.plot(metrics['metrics']['mAP_0.5:0.95'], label='mAP@0.5:0.95')
    plt.title('mAP during training')
    plt.xlabel('Epoch')
    plt.ylabel('mAP')
    plt.legend()
    
    plt.show()
else:
    print("Metrics not found. Utilisez les plots intégrés YOLO.")

# -------------------------------
# 3️⃣ Évaluation sur le jeu de validation
# -------------------------------
val_metrics = yolo_model.val(
    data='data.yaml',
    split='val',
    imgsz=960,
    batch=16
)
print("📊 Résultats de validation :")
print(val_metrics)

# -------------------------------
# 4️⃣ Fonction d'inference avec recalibration
# -------------------------------
def enlarge_bbox(xyxy, scale=1.3, W=0, H=0):
    x1, y1, x2, y2 = xyxy
    w = x2 - x1
    h = y2 - y1
    cx = (x1 + x2)/2
    cy = (y1 + y2)/2
    new_w = w*scale
    new_h = h*scale
    nx1 = max(0, cx - new_w/2)
    ny1 = max(0, cy - new_h/2)
    nx2 = min(W, cx + new_w/2)
    ny2 = min(H, cy + new_h/2)
    return [nx1, ny1, nx2, ny2]

transform = transforms.Compose([
    transforms.Resize((128,128)),
    transforms.ToTensor()
])

from ultralytics.engine.results import Results
import numpy as np

def infer_with_recalibration(img_path, save_dir="runs/test_predictions", conf_thresh=0.3):
    os.makedirs(save_dir, exist_ok=True)
    img = Image.open(img_path).convert("RGB")
    W, H = img.size
    
    results = yolo_model(img_path)[0]

    boxes = results.boxes.xyxy.cpu().numpy()
    classes = results.boxes.cls.cpu().numpy()
    scores  = results.boxes.conf.cpu().numpy()

    new_scores = scores.copy()

    for i, box in enumerate(results.boxes):
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        cls = int(box.cls[0])
        score = float(box.conf[0])

        crop = img.crop((x1, y1, x2, y2))
        crop_ctx = img.crop(enlarge_bbox([x1,y1,x2,y2],1.3,W,H))

        if os.path.exists(classifier_path):
            yolo_feat = transforms.ToTensor()(crop).view(-1)[:256]
            ctx_feat = ctx_encoder(transform(crop_ctx).unsqueeze(0)).squeeze(0)

            with torch.no_grad():
                scores_pred = classifier(yolo_feat.unsqueeze(0), ctx_feat.unsqueeze(0))
                new_score = torch.softmax(scores_pred, dim=1)[0][cls].item()
        else:
            new_score = score

        new_scores[i] = new_score

        print(f"Classe {cls} | Ancien score {score:.2f} → Nouveau score {new_score:.2f}")

    # --------------------------
    # 🔄 Créer un Results recalibré
    # --------------------------
    new_boxes_array = np.hstack([
        boxes,
        new_scores.reshape(-1,1),
        classes.reshape(-1,1)
    ])

    new_results = Results(
        orig_img=results.orig_img,
        path=results.path,
        names=results.names,
        boxes=torch.tensor(new_boxes_array),
        masks=None,
        probs=None
    )

    # --------------------------
    # 🔵 PLOT + SAVE (image annotée)
    # --------------------------
    annotated_img = new_results.plot()  # numpy array annotée
    annotated_pil = Image.fromarray(annotated_img)

    out_path = os.path.join(save_dir, os.path.basename(img_path))
    annotated_pil.save(out_path)

    print(f"📸 Image annotée sauvegardée dans : {out_path}")



# -------------------------------
# 5️⃣ Prédiction sur le jeu de test
# -------------------------------
#img_dir = os.path.join(base_dir, "Taco_Dataset/test/images")

#test_images = os.path.join(base_dir, 'Taco_Dataset/test/images/*.jpg')  
test_images = glob.glob('Taco_Dataset/test/images/*.jpg')  
print(f"📌 {len(test_images)} images trouvées pour test")

for img_path in test_images:
    infer_with_recalibration(img_path)

print(f"✅ Prédictions sauvegardées dans runs/test_predictions")
```
